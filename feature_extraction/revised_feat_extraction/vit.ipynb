```python
from PIL import Image, ImageSequence

def get_middle_frame(gif_path):
    """
    Get the middle frame of a GIF.

    Args:
    - gif_path (str): Path to the GIF file.

    Returns:
    - Image.Image: PIL Image object representing the middle frame.
    """
    with Image.open(gif_path) as im:
        frames = [frame.copy() for frame in ImageSequence.Iterator(im)]
        middle_index = len(frames) // 2
    return frames[middle_index]

# Test the function
temp_gif_location = '/workspace/data/data_folder/o2o/ac1VUY0wFkg_2867_5.gif'
temp_i3d_video = get_middle_frame(temp_gif_location)

import clip
clip_model, clip_preproc = clip.load("ViT-B/32")

preproc_image = clip_preproc(temp_i3d_video)
```
