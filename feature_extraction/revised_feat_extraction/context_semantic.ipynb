```python
import re

def split_camel_case(word):
    """
    Split camel case words into a list of individual words.
    
    Args:
    - word (str): A camel case word.
    
    Returns:
    - List of words split based on capital letters.
    """
    matches = re.findall('.+?(?:(?<=[a-z])(?=[A-Z])|(?<=[A-Z])(?=[A-Z][a-z])|$)', word)
    return matches

# Test
words = ['AssembleCabinet', 'ChangeCarTire', 'FuelCar']
for word in words:
    print(split_camel_case(word))
```

```python
import gensim.downloader as api
import numpy as np

# Load the Word2Vec model (this is a one-time process)
word2vec_model = api.load('word2vec-google-news-300')

def get_phrase_embedding(phrase, model):
    words = split_camel_case(phrase)
    vector_sum = sum([model[word] for word in words])  # Sum the embeddings of the words
    return vector_sum / len(words)  # Return the average

phrases = ['AssembleCabinet', 'ChangeCarTire', 'FuelCar', 'InstallBicycleRack', 'InstallShowerHead', 'ParkParallel', 'ReplaceDoorKnob', 'ReplaceToiletSeat', 'UseJack', 'PolishCar', 'ReplaceBatteryOnTVControl']
embeddings = {phrase: get_phrase_embedding(phrase, word2vec_model) for phrase in phrases}
```
